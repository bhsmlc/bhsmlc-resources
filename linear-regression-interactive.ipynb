## Defining our dataset

Let's use Python arrays to store our data

`data_y` represents the y values of the points, which is the amount of iced coffee sold in gallons

`(data_x[i], data_y[i])` are the coordinates of point i

```python
x_data = [70, 80, 75, 90, 65, 85]
y_data = [5.0, 8.2, 6.5, 11.5, 4.0, 10.0]

print(x_data)
print(y_data)
```

## Summing our data

The closed form formula for calculating the slope of the line of best fit involves finding the mean of the data. 

In order to find the mean, we must first find the sum. You can either do this with a built-in function or with an explicit way.

### Explicit Way

Using a for loop

```python
x_sum = 0
for num in x_data:
  x_sum += num
y_sum = 0
for num in y_data:
  y_sum += num

print(x_sum)
print(y_sum)
```

### Using a built-in funciton
You can also simply use the `sum()` function

```python
x_sum = sum(x_data)
y_sum = sum(y_data)

print(x_sum)
print(y_sum)
```

## Finding the mean

mean is the sum / length

To get the length of our data, we would use the `len()` function

```python
x_length = len(x_data)
y_length = len(y_data)

x_mean = x_sum / x_length
y_mean = y_sum / y_length

print(x_mean)
print(y_mean)
```

## Finding the slope of our line of best fit

Now that we have the necessary information, we can translate our math formula into code. The sigma would turn into a for loop.

Let's do this in 2 steps.

1. Finding the "rise," which is the numerator of the formula

2. Finding the "run," which is the denominator

### Finding the "rise"

Translating the below code to English, we are looking at each data point, and multiplying the x displacement from mean to the y displacement from mean and summing all those products.

```python
length = len(x_data)
rise = 0
for i in range(length):
  rise += (x_mean - x_data[i]) * (y_mean - y_data[i])
```

### Finding the "run"

Translating the below code to English, we find the squared error from each x value to the x mean and we sum the errors.

```python
length = len(x_data)
run = 0
for i in range(length):
  run += (x_mean - x_data[i])**2
```

### Calculate slope (m) with rise/run

```python
m = rise / run
print(m)
```

## Find the y-intercept using the slope

We'll use the formula

y_mean = m*x_mean+b

which rearranges to
b = y_mean - m*x_mean

```python
b = y_mean - m*x_mean
print(b)
```

## Create the function of the line of best fit

This will allow you to input an x value and get a predicted y value

```python
def f(x):
  x = min(120, x) # any temperature over 120 is unrealistic so it will be taken as 120
  prediction = m*x+b
  return max(0, prediction) # you can't sell negative coffee

print("f(x) = "+str(m)+"x"+" + "+str(b))
```

## Make some predictions

Now that we have out function, we can make some predictions with it

```python
print(f(2)) # Prediction of gollons of coffee sold in 2 degrees F
print(f(20)) # Prediction of gollons of coffee sold in 20 degrees F
print(f(80)) # Prediction of gollons of coffee sold in 80 degrees F
print(f(90)) # Prediction of gollons of coffee sold in 90 degrees F
print(f(105)) # Prediction of gollons of coffee sold in 105 degrees F

```
