```python
import matplotlib.pyplot as plt
from sklearn.preprocessing import OrdinalEncoder
from sklearn.tree import DecisionTreeClassifier, plot_tree

# 1. Creating training data
x_train = [
    ["Sunny", "Hot"],
    ["Sunny", "Mild"],
    ["Rainy", "Mild"],
    ["Cloudy", "Mild"],
    ["Cloudy", "Hot"],
]
y_train = ["No", "Yes", "No", "Yes", "Yes"]

# 2. Creating test data
x_test = [[i, j] for i in ["Sunny", "Rainy", "Cloudy"] for j in ["Hot", "Mild"]]
y_test = ["No", "Yes", "No", "No", "Yes", "Yes"]

# 3. Encoding the strings to numbers
encoder = OrdinalEncoder(
    categories=[
        ["Sunny", "Rainy", "Cloudy"],
        ["Hot", "Mild", "Cold"]
    ]
)
encoded_x_train = encoder.fit_transform(x_train)
encoded_x_test = encoder.transform(x_test)

# 4. Training the model
clt = DecisionTreeClassifier(max_depth=3, random_state=42)
clt.fit(encoded_x_train, y_train)

# 5. Visualizing the tree
plt.figure(figsize=(12, 8))
plot_tree(
    clt,
    feature_names=["Outlook", "Temperature"], # Outlook is rainy/cloudy/sunny, temp is hot/mild
    class_names=["No", "Yes"],
    filled=True,
)
plt.show()

# 6. Evaluating
y_pred = clt.predict(encoded_x_test)
accuracy = clt.score(encoded_x_test, y_test)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
```
